```python
from crewai import Crew,Agent,Task,LLM
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)
```

```python
llm=LLM(
    api_key=os.getenv("MISTRAL_API_KEY"),
    model="open-mixtral-8x7b",
    base_url="https://api.mistral.ai/v1",
)
print("Mistral LLM connected successfully!")
```

```python


agent=Agent(
    role="Research Agent",
    goal="Research the user's question and provide a clear and accurate answer.",
    backstory="You are a helpful research assistant who explains information in simple language.",
    llm=llm,
    verbose=True
)
```

```python
task = Task(
    description="""
    Research the following user question and provide a clear,
    accurate, and easy-to-understand answer:

    {question}
    """,
    expected_output="""
    A clear and accurate answer to the user's question.
    Use simple language and organize the answer logically.
    """,
    agent=agent,
)
```

```python
from crewai import Crew

crew = Crew(
    agents=[agent],
    tasks=[task],
    verbose=True,
)
```

```python
question = input("Enter your research question: ")
```

```python
result=await crew.kickoff_async(
    inputs={
        "question":question
    }
)
```

```python
from IPython.display import Markdown,display

display(Markdown(result.raw))
```
