```python
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent,Runner,OpenAIChatCompletionsModel,function_tool
from IPython.display import Markdown,display
import os
from pydantic import BaseModel
```

```python
load_dotenv(override=True)
```

```python
client=AsyncOpenAI(
    api_key=os.getenv("OPENROUTER_API_KEY_2"),
    base_url="https://openrouter.ai/api/v1"
)
```

```python
model=OpenAIChatCompletionsModel(
    model="openai/gpt-oss-20b:free",
    openai_client=client
)
```

```python
class StuctureOutput(BaseModel):
    is_Question:str
    is_answer:str
```

```python
@function_tool
def add_tool(num1:float,num2:float)->float:
    "You are the add tool your responsiblity is to add two numbers"
    return num1+num2
```

```python
@function_tool
def sub_tool(num1:float,num2:float)->float:
    "You are the substract tool your responsiblity is to sunstract two numbers"
    return num1-num2
```

```python
@function_tool
def multiply_tool(num1:float,num2:float)->float:
    "You are the multiply tool your responsiblity is to multiply two numbers"
    return num1*num2
```

```python
@function_tool
def divid_tool(num1:float,num2:float)->float:
    "You are the divid tool your responsiblity is to divid two numbers"
    if num1==0:
        return 0
    return num1/num2
```

```python
calculator_agent=Agent(
    name="Calculator bot",
    instructions="""
    you are the Calculator bot your responsibility is to answer only add , substract , multiply and divid question by using your connected tool:
    - add_tool(num1,num2)
    - sub_tool(num1,num2)
    - multiply_tool(num1,num2)
    - divid_tool(num1,num2)
    """,
    model=model,
    tools=[add_tool,sub_tool,multiply_tool,divid_tool],
    
)
```

```python
response=await Runner.run(
    calculator_agent,"divid two numbers 2/2"
)
```

```python
display(Markdown(response.final_output))
```
