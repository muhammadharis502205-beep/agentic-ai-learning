```python
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent,Runner,OpenAIChatCompletionsModel,function_tool
import os
from ddgs import DDGS
from IPython.display import Markdown,display
```

```python
import sys
print(sys.executable)
```

```python
load_dotenv(override=True)
```

```python
client=AsyncOpenAI(
    api_key=os.getenv("OPENROUTER_API_KEY_1"),
    base_url="https://openrouter.ai/api/v1"
)
```

```python
model=OpenAIChatCompletionsModel(
    model="openai/gpt-oss-20b:free",
    openai_client=client
)
```

```python
@function_tool
def search_tool(query)->str:
    """
    Searches the web for a given query and returns matching results 
(title, url, and snippet) to ground the agent's answer in 
real, cited information.
    """
    with DDGS() as ddgs:
        results=ddgs.text(query,max_results=4)

        if not results:
            return "Result not found"
        output=[]
        for result in results:
            output.append(
            f"Title: {result['title']}\n"
            f"URL: {result['href']}\n"
            f"Snippet: {result['body']}"
        )

    return "\n\n".join(output)
```

```python

research_agent=Agent(
    name="Deep Research Agent",
    instructions="""
    You are a Deep Research Agent.

RULE: Never answer from your own knowledge. For every user question, 
you MUST call search_tool(query) first and build your answer only 
from its results.

WORKFLOW:
1. Read the user's question.
2. Call search_tool(query) with a short, focused query (break complex 
   questions into 2-3 sub-queries if needed, call search_tool for each).
3. Wait for results.
4. Write a clear, well-structured summary using ONLY the returned content.
5. For every claim, attach the source link inline as: [Title](href)

OUTPUT FORMAT:
## Summary
[2-4 sentence overview of the answer]

## Details
- Point 1 — with explanation [Source Title](href)
- Point 2 — with explanation [Source Title](href)
- ...

## Sources
- [Title 1](href1)
- [Title 2](href2)

CONSTRAINTS:
- If search_tool returns nothing relevant, say so honestly — do not guess.
- If search_tool fails, returns an error, or gives no response at all, 
  you MUST NOT answer from your own knowledge under any circumstances. 
  Simply tell the user that search results are unavailable right now, 
  and stop there.
- Keep language simple and skimmable — short paragraphs, bullet points, bold key terms.
- Never fabricate a link or title not returned by search_tool.
- Always cite the specific source next to the specific claim, not just at the end.
    """,
    model=model,
    tools=[search_tool],

)
```

```python
response=await Runner.run(
    research_agent,"give me latest news on autonomius agentic ai"
)
```

```python
display(Markdown(response.final_output))
```
