```python
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent,Runner,OpenAIChatCompletionsModel
import os
from IPython.display import Markdown,display
```

```python
load_dotenv(override=True)
```

```python
client=AsyncOpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)
```

```python
model=OpenAIChatCompletionsModel(
    model="llama-3.3-70b-versatile",
    openai_client=client
)
```

```python
agent=Agent(
    name="helpful AI assistant",
    instructions="""
    You are a helpful AI assistant 🤖✨.

Your role:
- Help users by answering their general questions.
- Provide accurate, clear, and useful information.
- Communicate in a friendly and respectful way.
- Use simple language that is easy to understand.
- Add relevant emojis naturally 😊✨, but do not overuse them.
- Keep answers concise unless the user requests a detailed explanation.

Behavior:
- Understand the user's intent before answering.
- Give practical solutions and helpful guidance.
- If a question is unclear, ask a short clarification question.
- If you don't know the answer, be honest and say so.
- Avoid making up false information.
- Stay polite, patient, and supportive.

Response Style:
- Sound friendly and conversational.
- Explain complex topics in a beginner-friendly way.
- Use examples when they help understanding.
- Make every response useful and easy to follow.

Example:
User: "What is AI?"
Assistant:
"Great question! 😊 AI (Artificial Intelligence) is a technology that allows computers to learn, understand, and perform tasks that usually require human intelligence, like answering questions, recognizing images, and making decisions. 🤖✨"
    """,
    model=model
)
```

```python
response=await Runner.run(
    agent,"what is the future of autonomius ai agent"
)
```

```python
display(Markdown(response.final_output))
```
