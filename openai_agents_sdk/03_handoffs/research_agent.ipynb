```python
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent,Runner,OpenAIChatCompletionsModel,handoff,function_tool
import os
from IPython.display import Markdown,display
from ddgs import DDGS
```

```python
load_dotenv(override=True)
```

```python
client=AsyncOpenAI(
    api_key=os.getenv("OPENROUTER_API_KEY_2"),
    base_url="https://openrouter.ai/api/v1"
)
```

```python
model=OpenAIChatCompletionsModel(
    model="openai/gpt-oss-20b:free",
    openai_client=client
)
```

```python
@function_tool
def search_tool(query:str)->str:
    """
    Search the web for up-to-date information about a given query and return relevant results for research.
    """
    with DDGS() as ddgs:
        results=ddgs.text(query,max_results=3)
    if not results:
        return f"No results found for '{query}'."
 
    formatted = []
    for i, r in enumerate(results, start=1):
        title = r.get("title", "")
        body = r.get("body", "")
        href = r.get("href", "")
        formatted.append(f"{i}. {title}\n   {body}\n   Source: {href}")
 
    return "\n\n".join(formatted)
```

```python
writer_agent = Agent(
    name="Writer Agent",
    instructions=(
        """You are a Writer Agent. Your task is to convert structured notes into a well-formatted, polished report.

INPUT: You will receive structured notes (bullet points, headings, or raw data).

OUTPUT: A clean, professional report in Markdown format.

RULES:
1. Preserve all factual content from the notes — do not omit, invent, or alter information.
2. Organize content under clear sections/headings based on the input structure.
3. Convert bullet points into well-written paragraphs where appropriate, but keep lists where they aid readability (steps, comparisons, data points).
4. Use a professional, neutral tone unless the notes indicate otherwise.
5. Add a short title and (if content is long) a brief 2-3 line summary at the top.
6. Keep formatting consistent: one heading style, one list style, consistent spacing.
7. Do not add opinions, extra commentary, or filler content not present in the notes.
8. If the notes are incomplete or ambiguous, note it briefly instead of guessing.

GOAL: Turn raw notes into a report that is clear, scannable, and ready to share — without changing the meaning of the original content."""
    ),
    model=model
)
 
```

```python
research_agent = Agent(
    name="Research Agent",
    instructions=(
        """You are a Research Agent.

Your job is to research the user's query using the `search_tool(query)`.

Rules:
- Always use `search_tool` for research.
- Never answer from your own knowledge.
- Base your output only on search results.
- Extract the most relevant and useful information.
- Convert the findings into clear, structured notes.
- Keep notes concise and well-organized.
- Do not write the final report."""
    ),
    tools=[search_tool],
    model=model
)
```

```python
user_query = "What are the latest developments in solid-state batteries in 2026?"
 
result = await Runner.run(research_agent, user_query)
```

```python
response=await Runner.run(
    writer_agent,result.final_output
)
```

```python
display(Markdown(response.final_output))
```
